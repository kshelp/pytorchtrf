```python
######################
# 1. 탐색한 하이퍼 파라미터를 훈련에 적용하기
from ultralytics import YOLO
import shutil
import yaml

# YOLOv8 모델 로드
model = YOLO("yolov8n.pt")

# 하이퍼파라미터 yaml 파일 불러오기
with open("/content/drive/MyDrive/models/best_hyperparameters.yaml", "r", encoding="utf-8") as f:
    hyp_dict = yaml.safe_load(f)

# 모델 훈련
#model.train(data="/content/drive/MyDrive/datasets/numbers-50/data.yaml", epochs=250, imgsz=640, **hyp_dict)

# 학습 (결과 객체 받기)
results = model.train(
    data="/content/drive/MyDrive/datasets/numbers-50/data.yaml",
    epochs=250,
    imgsz=640,
    project="/content/drive/MyDrive/run",  # 저장 위치
    name="numbers_model",  # 16행
    **hyp_dict
)

# best 모델 경로
best_model_path = "/content/drive/MyDrive/run/numbers_model/weights/best.pt"
last_model_path = "/content/drive/MyDrive/run/numbers_model/weights/last.pt"

print(f"Best model saved at: {best_model_path}")
print(f"Last model saved at: {last_model_path}")

# 원하는 위치로 복사 (선택)
save_path = "/content/drive/MyDrive/models/numbers_best.pt"
shutil.copy(best_model_path, save_path)

print(f"Copied best model to: {save_path}")
```
