```python
######################
# 1. 파라미터 탐색하기
from ultralytics import YOLO

# YOLOv8 모델 로드
model = YOLO("yolov8n.pt")

# 하이퍼파라미터 탐색 수행
model.tune(
    data="/content/drive/MyDrive/datasets/numbers-50/data.yaml",
    epochs=20,
    iterations=25, # 탐색 횟수 (예: 25번의 실험)
    val=True
)
```
