```python
######################
# 1. 커스텀 데이터로 학습시키기
from ultralytics import YOLO
import cv2
import shutil

# 모델 로드
model = YOLO("yolov8n.pt")

# 학습 (결과 객체 받기)
results = model.train(
    data="/content/drive/MyDrive/datasets/numbers-50/data.yaml",
    epochs=250,
    imgsz=640,
    project="/content/drive/MyDrive/run",  # 저장 위치
    name="numbers_model"  # 폴더 이름
)

# best 모델 경로
best_model_path = "/content/drive/MyDrive/run/numbers_model/weights/best.pt"
last_model_path = "/content/drive/MyDrive/run/numbers_model/weights/last.pt"

print(f"Best model saved at: {best_model_path}")
print(f"Last model saved at: {last_model_path}")

# 원하는 위치로 복사 (선택)
save_path = "/content/drive/MyDrive/models/numbers_best.pt"
shutil.copy(best_model_path, save_path)

print(f"Copied best model to: {save_path}")

# 저장된 모델 로드해서 사용 (권장)
model = YOLO(best_model_path)

img_path = '/content/drive/MyDrive/datasets/number.jpg'
result = model.predict(img_path)

print(result[0].boxes)

# 이미지 로드
img = cv2.imread(img_path)

# GPU → CPU → numpy 변환
boxes = result[0].boxes.xyxy.cpu().numpy()
classes = result[0].boxes.cls.cpu().numpy()
confs = result[0].boxes.conf.cpu().numpy()

# 박스 그리기
for i in range(len(boxes)):
    x1, y1, x2, y2 = boxes[i].astype(int)
    cls = int(classes[i])
    conf = confs[i]

    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 0, 255), 2)
    cv2.putText(
        img,
        f'{cls} {conf:.4f}',
        (x1, y1 - 5),
        cv2.FONT_HERSHEY_PLAIN,
        2,
        (0, 0, 255),
        2
    )

# 파일로 저장
output_path = "/content/drive/MyDrive/output/number_result.jpg"
cv2.imwrite(output_path, img)
print(f"Saved result to: {output_path}")

# 화면에 표시
# cv2.imshow('image', img)
# cv2.waitKey()
# cv2.destroyAllWindows()
```
