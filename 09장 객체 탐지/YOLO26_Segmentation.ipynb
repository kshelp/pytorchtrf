```python
######################
# 1.
from ultralytics import YOLO
import cv2
import numpy as np

# 모델 불러오기
model = YOLO("yolo26n-seg.pt")

# 추론 실행
results = model("/content/drive/MyDrive/datasets/sample.png", conf=0.25, iou=0.45)

# 결과
result = results[0]
print(f"Objects detected: {len(result.boxes)}")

# 이미지 가져오기
original = result.orig_img
annotated = result.plot()

# 좌/우 붙이기 (가로 concat)
combined = np.hstack((original, annotated))

# 크기 줄이기 (예: 50%)
scale = 0.5
combined_resized = cv2.resize(
    combined,
    (2*int(combined.shape[1]*scale), int(combined.shape[0]*scale))
)

# 창 설정 (크기 조절 가능)
cv2.namedWindow("Result", cv2.WINDOW_NORMAL)
cv2.imshow("Result", combined_resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
